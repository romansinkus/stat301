```r
library(tidyverse)
library(moderndive)
library(car)
```

```r
wage_data <- read.table("data/wage.txt", header=TRUE)
wage_data$wage_categorical <- as.numeric(wage_data$wage > mean(wage_data$wage))
head(wage_data)
```

### Part A

```r
wage_logistic_model <- glm(formula = wage_categorical ~ education + sex, data=wage_data, family='binomial')
summary(wage_logistic_model)
```

EDUCATION and SEX are both associated with an individual's wage being above or below average. We can observe `Pr(>|z|)` which indicates the p-value is `1.41e-13` for EDUCATION and `2.89e-05` for SEX. Both of these values are significantly below 0.05 which indicates that the two coefficiencts are statistically significant and therefore there is enough evidence to claim that the coefficients are not 0.

Increased EDUCATION is associated with higher odds of above average wage (`0.30942`). Being female is associated with lower odds of having an above average wage (`-0.81518`) and therefore being a male is associated with higher odds of having an above average wage.

Note: for this dataset, 0=male and 1=female.

### Part B

```r
l_i <- (-4.15150) + (-0.81518) + 16*(0.30942)
estimated_prob <- exp(l_i)/(1+exp(l_i))
estimated_prob
```

There is a `49.60 %` probability that a female with 16 years of education has a wage that is above average.
