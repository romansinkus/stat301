```r
library(tidyverse)
library(moderndive)
library(car)
```

```r
wage_data <- read.table("data/wage.txt", header=TRUE)
head(wage_data)
```

### Part A

```r
model_1 <- lm(wage ~ education, data = wage_data)
model_2 <- lm(wage ~ education + sex, data = wage_data)
summary(model_1)$adj.r.squared
summary(model_2)$adj.r.squared
```

When comparing the two models, model 2 appears to have a higher proportion of total variance of the response explained by the model while penalizing the addition of predictors (i.e. it exhibits a higher adjusted R^2 value) compared to model 1. More specifically, model 1 has an adjusted R^2 value of `0.144` whereas model 2 has an adjusted R^2 value of `0.185`. Given that we are examining the adjusted R^2 value and not the raw R^2 value, this increase is not simply attributed to adding additional variables to the model. Therefore, model 2 is a better model. Note, this does not tell us anything about the significance of adding the sex predictor to the model.

### Part B

```r
anova(model_1, model_2)
```

In the table above, model 1 is the reduced model whereas model 2 is the full model (which includes the additional predictor sex). The F-test compares the nested models by determining if the sex predictor is significant.

According to the table above, adding model 2 reduces the Sum of Squares by `598.2` which indicates that model 2 is explaining more of the variance in the data compared to model 1.

The p-value in the table above is `1.95e-07` which is significantly lower than the `0.05` significance level. Therefore at a `0.05` significance level, we reject the null hypothesis that the sex predictor has no effect on the model.

Therefore, model 2 is significantly better than model 1.
