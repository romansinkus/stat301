```r
library(tidyverse)
library(moderndive)
library(car)
```

```r
hiv_data <- read.table("data/hiv_data.txt", head=TRUE)
head(hiv_data)
```

### Part A

```r
poisson_model <- glm(CD4 ~ trarm + age + weight + logRNA, family=poisson, data=hiv_data)
summary(poisson_model)
```

The p-value for trarm is quite loww (`2e-16`) and is therefore below the 0.05 significance level. Therefore at a 0.05 significance level, we reject the null hypothesis that B=0. Therefore, the treatment arm is statistically significant for determining CD4 levels.

### Part B

```r
poisson_model <- glm(CD4 ~ trarm + age + weight + logRNA, family = quasipoisson, data=hiv_data)
summary(poisson_model)
```

The disersion parameter for the poisson family is `70.09` when fitting the model using the quasilikelihood method. Therefore, overdispersion is a problem for the model above since it's quite far from 1.
