```r
library(tidyverse)
library(moderndive)
library(car)
```

```r
wage_data <- read.table("data/wage.txt", header=TRUE)
head(wage_data)
```

### Part A

```r
wage_data$wage_categorical <- as.numeric(wage_data$wage > mean(wage_data$wage))
head(wage_data)
```

```r
wage_logistic_model <- glm(formula = wage_categorical ~ education, data=wage_data, family='binomial')
summary(wage_logistic_model)
```

`Education` is associated with an individual's wage being above/below average given the very small p-value `1.86e-13`. The `0.3017` estimate indicates that each additional year of education is associated with multiplying the odds (of being above average) by `e^0.3017 = 1.35`. This means that each additional year of education is associated with a 35% increase in the odds that the wage will be above average.
